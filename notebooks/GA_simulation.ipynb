```python
# import final parameters and run simulation


from llm_abm_ga.logging_utils import setup_logger

log_file = setup_logger("results/logs")
print("Logging to:", Path(log_file).resolve())

from llm_abm_ga.ga.evolution import evolve_prompts
from llm_abm_ga.prompts.components import system_template_phrases, user_template_phrases

# Running the GA
population = evolve_prompts(
    system_template_phrases,
    user_template_phrases,
    n_generations=15,
    pop_size=50,
    n_buyers=80000,
    periods=10,
    tournament_k=6,
    elite_size=5,
    mutation_rate=0.1,
    log_file=log_file,
)

print("GA completed.")
```
